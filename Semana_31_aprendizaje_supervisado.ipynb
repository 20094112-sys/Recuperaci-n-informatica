```python
from sklearn.neighbors import KNeighborsClassifier

# Datos de entrenamiento
X = [
    [1, 1],
    [2, 1],
    [4, 4],
    [5, 4]
]

# Categorías
y = ["Pequeña", "Pequeña", "Grande", "Grande"]

# Crear y entrenar el modelo
modelo = KNeighborsClassifier(n_neighbors=1)
modelo.fit(X, y)

# Realizar una predicción
prediccion = modelo.predict([[1.5, 1]])

print("La flor pertenece a la categoría:")
print(prediccion[0])
```

```python
from sklearn.tree import DecisionTreeClassifier

# Datos de entrenamiento
X = [
    [1, 1],
    [2, 1],
    [4, 4],
    [5, 4]
]

# Categorías
y = ["Pequeña", "Pequeña", "Grande", "Grande"]

# Crear y entrenar el árbol
arbol = DecisionTreeClassifier(random_state=42)
arbol.fit(X, y)

# Realizar una predicción
resultado = arbol.predict([[5, 5]])

print("Clasificación de la flor:")
print(resultado[0])
```

```python
from sklearn.metrics import accuracy_score

# Datos para evaluar el modelo
X_prueba = [
    [1, 1],
    [2, 1],
    [4, 4],
    [5, 4]
]

y_real = ["Pequeña", "Pequeña", "Grande", "Grande"]

# Realizar predicciones
y_predicho = arbol.predict(X_prueba)

# Calcular precisión
precision = accuracy_score(y_real, y_predicho)

print("Predicciones:", y_predicho)
print("Precisión del modelo:", precision * 100, "%")
```
