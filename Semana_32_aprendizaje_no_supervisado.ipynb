```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Crear datos de ejemplo
X = [
    [1, 2], [2, 1], [1, 1],
    [8, 8], [9, 8], [8, 9],
    [4, 5], [5, 4], [5, 5]
]

# Crear el modelo K-Means
modelo = KMeans(n_clusters=3, random_state=42, n_init=10)

# Entrenar el modelo
modelo.fit(X)

# Obtener los grupos
grupos = modelo.labels_

print("Grupos encontrados:")
print(grupos)

# Mostrar los grupos en una gráfica
plt.scatter(
    [p[0] for p in X],
    [p[1] for p in X],
    c=grupos
)

plt.title("Agrupación de datos con K-Means")
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Probar diferentes cantidades de grupos
errores = []

for k in range(1, 7):
    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    modelo.fit(X)
    errores.append(modelo.inertia_)

# Crear gráfica del método del codo
plt.plot(range(1, 7), errores, marker="o")
plt.title("Método del codo")
plt.xlabel("Cantidad de grupos")
plt.ylabel("Error")
plt.xticks(range(1, 7))
plt.show()
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Reducir los datos a dos dimensiones
pca = PCA(n_components=2)
datos_reducidos = pca.fit_transform(X)

# Mostrar los datos en una gráfica
plt.scatter(
    datos_reducidos[:, 0],
    datos_reducidos[:, 1]
)

plt.title("Visualización de datos con PCA")
plt.xlabel("Componente principal 1")
plt.ylabel("Componente principal 2")
plt.show()

print("Datos originales:", len(X), "filas y", len(X[0]), "columnas")
print("Datos reducidos:", datos_reducidos.shape)
```
