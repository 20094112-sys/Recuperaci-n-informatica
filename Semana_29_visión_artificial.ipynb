```python
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files

# Subir una imagen
archivo = files.upload()

# Abrir la imagen
nombre = list(archivo.keys())[0]
imagen = Image.open(nombre)

# Mostrar la imagen
plt.imshow(imagen)
plt.title("Mi primera imagen en visión artificial")
plt.axis("off")
plt.show()

# Mostrar tamaño de la imagen
print("Tamaño de la imagen:", imagen.size)
```

```python
# Convertir la imagen a escala de grises
imagen_gris = imagen.convert("L")

# Mostrar la imagen
plt.imshow(imagen_gris, cmap="gray")
plt.title("Imagen en escala de grises")
plt.axis("off")
plt.show()
```

```python
import numpy as np
from PIL import ImageFilter

# Detectar los bordes de la imagen
bordes = imagen.convert("L").filter(ImageFilter.FIND_EDGES)

# Mostrar el resultado
plt.imshow(bordes, cmap="gray")
plt.title("Detección de bordes")
plt.axis("off")
plt.show()
```
